```python
import re
import pandas as pd


pat_ft = re.compile(r'Full-time\s+(\d+)\s+months')
pat_pt = re.compile(r'Part-time\s+(\d+)\s+months')

def to_mode_fields(txt):
    if pd.isna(txt):
        return pd.NA, pd.NA
    s = str(txt)

    modes = []
    months = []


    m_ft = pat_ft.search(s)
    if m_ft:
        modes.append("Full-time")
        months.append(m_ft.group(1))

    m_pt = pat_pt.search(s)
    if m_pt:
        modes.append("Part-time")
        months.append(m_pt.group(1))

    if not modes:
        return pd.NA, pd.NA

    return " | ".join(modes), " | ".join(months)

df[['study_mode', 'study_mode_months']] = df['durations'].apply(
    lambda x: pd.Series(to_mode_fields(x))
)
```

```python
import re
import pandas as pd
import numpy as np

df['start_dates'] = df['start_dates'].astype(str).str.replace("Apply before Deadline", "Apply before", regex=False)

MONTH_MAP = {
    'January':'01','Jan':'01',
    'February':'02','Feb':'02',
    'March':'03','Mar':'03',
    'April':'04','Apr':'04',
    'May':'05',
    'June':'06','Jun':'06',
    'July':'07','Jul':'07',
    'August':'08','Aug':'08',
    'September':'09','Sep':'09',
    'October':'10','Oct':'10',
    'November':'11','Nov':'11',
    'December':'12','Dec':'12'
}

RE_START = re.compile(r'^Starting\s+([A-Za-z]+)\s+(\d{4})$')
RE_APPLY = re.compile(r'^Apply before\s+([A-Za-z]+)\s+(\d{4})$')

def to_mm_yyyy(month_str, year_str):
    mm = MONTH_MAP.get(month_str, None)
    if mm is None:
        return pd.NA
    return f"{mm}.{year_str}"

def split_all_pairs(text):
    """
    - Metni '|' ile böler, sırayla tüm 'Starting' ve 'Apply before' bölümlerini toplar.
    - 'Starting'ler intake listesine, 'Apply before'lar deadline listesine eklenir.
    - Apply sayısı Starting sayısından fazlaysa Starting sayısına kadar sınırlandırılır.
    - Çıktı iki metin: "MM.YYYY | MM.YYYY ..." biçiminde.
    """
    if pd.isna(text) or str(text).strip() == "" or text == "nan":
        return pd.NA, pd.NA

    parts = [p.strip() for p in str(text).split('|') if p.strip()]

    intake_list = []
    deadline_list = []

    for p in parts:
        ms = RE_START.match(p)
        if ms:
            intake_list.append(to_mm_yyyy(ms.group(1), ms.group(2)))
            continue
        ma = RE_APPLY.match(p)
        if ma:
            deadline_list.append(to_mm_yyyy(ma.group(1), ma.group(2)))
            continue

    if len(deadline_list) > len(intake_list):
        deadline_list = deadline_list[:len(intake_list)]

    intake_val = " | ".join([x for x in intake_list if pd.notna(x)]) if any(pd.notna(x) for x in intake_list) else pd.NA
    deadline_val = " | ".join([x for x in deadline_list if pd.notna(x)]) if any(pd.notna(x) for x in deadline_list) else pd.NA

    return intake_val, deadline_val

df[['intake_month', 'application_deadline_month']] = df['start_dates'].apply(
    lambda x: pd.Series(split_all_pairs(x), index=['intake_month', 'application_deadline_month'])
)
```

```python
import pandas as pd
import re
import unicodedata

LANG_MAP = {
    'english': 'English',
    'german': 'German',
    'russian': 'Russian',
    'spanish': 'Spanish',
    'french': 'French',
    'italian': 'Italian',
}


LANG_PATTERN = re.compile(r'\b(english|german|russian|spanish|french|italian)\b', flags=re.IGNORECASE)

def normalize_text(s):
    if pd.isna(s):
        return s
    s = unicodedata.normalize('NFKC', str(s)).strip()
    s = re.sub(r'\s+', ' ', s)
    return s

def clean_language_value(text):
    if pd.isna(text):
        return pd.NA
    s = normalize_text(text)

    found = []
    for m in LANG_PATTERN.finditer(s):
        k = m.group(1).lower()
        if k not in found:
            found.append(k)

    if not found:
        return pd.NA

    found = found[:2]
    return " | ".join(LANG_MAP[k] for k in found)

df['language_clean'] = df['language'].apply(clean_language_value)
```

```python
def clean_lang(text):
    if pd.isna(text):
        return ""
    s = str(text).strip()
    low = s.lower()

    out_parts = []

    def find_with_number(patterns):
        for pat in patterns:
            m = re.search(pat, low, flags=re.I)
            if m:
                num = m.group(m.lastindex) if m.lastindex else None
                if num:
                    return num.strip()
        return None

    # IELTS
    ielts_num = find_with_number([r"\bielts[^0-9]{0,15}(\d+(?:\.\d+)?)"])
    if ielts_num:
        out_parts.append(f"IELTS {ielts_num}")

    # TOEFL® IBT
    toefl_num = find_with_number([
        r"\btoefl\s*(?:i\s*bt|ibt|i-?bt)?[^0-9]{0,15}(\d+(?:\.\d+)?)",
        r"\bibt[^0-9]{0,15}(\d+(?:\.\d+)?)"
    ])
    if toefl_num:
        out_parts.append(f"TOEFL® IBT {toefl_num}")

    # PTE Academic
    pte_num = find_with_number([r"\bpte(?:\s*academic|\s*acad\.?)?[^0-9]{0,15}(\d+(?:\.\d+)?)"])
    if pte_num:
        out_parts.append(f"PTE Academic {pte_num}")

    # Duolingo English Test
    det_num = find_with_number([
        r"\bduolingo(?:\s*english\s*test)?[^0-9]{0,15}(\d+(?:\.\d+)?)",
        r"\bdet[^0-9]{0,15}(\d+(?:\.\d+)?)"
    ])
    if det_num:
        out_parts.append(f"Duolingo English Test {det_num}")

    # OTE
    ote_num = find_with_number([
        r"\boxford\s*test\s*of\s*english[^0-9]{0,15}(\d+(?:\.\d+)?)",
        r"\bote[^0-9]{0,15}(\d+(?:\.\d+)?)"
    ])
    if ote_num:
        out_parts.append(f"OTE {ote_num}")

    return " | ".join(out_parts).strip()

df["language_requirements_std"] = df["language_requirements"].apply(clean_lang)


print(df[["language_requirements", "language_requirements_std"]].head(12).to_string(index=False))
```

```python
def clean_delivery_type(text):
    if pd.isna(text):
        return ""
    s = str(text).lower().strip()

    oncampus = bool(re.search(r"\bon\s*campus\b", s))
    blended = bool(re.search(r"\bblended\b", s))
    online = bool(re.search(r"\bonline\b", s))

    parts = []
    if oncampus:
        parts.append("On Campus")
    if blended:
        parts.append("Blended")
    if online:
        parts.append("Online")

    return " | ".join(parts)

df["delivery_type_std"] = df["delivery_type"].apply(clean_delivery_type)

print(df[["delivery_type", "delivery_type_std"]].head(12).to_string(index=False))
```

```python
def clean_disciplines(cell):
    if pd.isna(cell):
        return ""
    s = str(cell).strip()

    s = re.sub(r"\|?\s*View.*", "", s, flags=re.I).strip()

    parts = [p.strip() for p in s.split("|") if p.strip()]

    seen = set()
    unique_parts = []
    for p in parts:
        if p not in seen:
            seen.add(p)
            unique_parts.append(p)

    return " | ".join(unique_parts)


df["disciplines_clean"] = df["disciplines"].apply(clean_disciplines)


print(df[["disciplines", "disciplines_clean"]].head(15).to_string(index=False))
```

```python
amt_pat = re.compile(r'"original_amount"\s*:\s*"(?P<amt>\d+(?:\.\d+)?)"')
cur_pat = re.compile(r'"original_currency"\s*:\s*"(?P<cur>[A-Z]{3})"')

def extract_tuition_regex(cell):
    if pd.isna(cell):
        return pd.Series({"tuition_amount": None, "tuition_currency": None})
    s = str(cell)
    m_amt = amt_pat.search(s)
    m_cur = cur_pat.search(s)
    amt = float(m_amt.group("amt")) if m_amt else None
    cur = m_cur.group("cur") if m_cur else None
    return pd.Series({"tuition_amount": amt, "tuition_currency": cur})

tuition_info = df["tuition"].apply(extract_tuition_regex)
df = pd.concat([df, tuition_info], axis=1)

print(df[["tuition", "tuition_amount", "tuition_currency"]].head(15).to_string(index=False))
```

```python
exchange_rates = {
    "EUR": 1.0,
    "GBP": 1.17,   # İngiliz Sterlini
    "USD": 0.87,   # ABD Doları
    "CHF": 0.93,   # İsviçre Frangı
    "SEK": 0.088,  # İsveç Kronu
    "NOK": 0.085,  # Norveç Kronu
    "DKK": 0.133,  # Danimarka Kronu
    "PLN": 0.23,   # Polonya Zlotisi
    "CZK": 0.042,  # Çek Korunası
    "HUF": 0.0025, # Macar Forinti
    "RON": 0.20,   # Romen Leyi
    "RUB": 0.011,  # Rus Rublesi
    "TRY": 0.02,   # Türk Lirası
    "ISK": 0.0069  # İzlanda Kronası
}

def convert_to_eur(row):
    amt = row.get("tuition_amount")
    cur = row.get("tuition_currency")
    if pd.isna(amt) or cur not in exchange_rates:
        return None
    return round(amt * exchange_rates[cur], 2)

df["tuition_amount_eur"] = df.apply(convert_to_eur, axis=1)

print(df[["tuition_amount", "tuition_currency", "tuition_amount_eur"]].head(15).to_string(index=False))
```

```python
min_pat = re.compile(r'"amount_min_base"\s*:\s*"(?P<min>\d+(?:\.\d+)?)"')
max_pat = re.compile(r'"amount_max_base"\s*:\s*"(?P<max>\d+(?:\.\d+)?)"')
cur_pat = re.compile(r'"base_currency"\s*:\s*"(?P<cur>[A-Z]{3})"')

def extract_living_info(cell):
    if pd.isna(cell):
        return pd.Series({"living_cost_min": None, "living_cost_max": None, "living_cost_currency": None})
    s = str(cell)
    m_min = min_pat.search(s)
    m_max = max_pat.search(s)
    m_cur = cur_pat.search(s)
    min_val = float(m_min.group("min")) if m_min else None
    max_val = float(m_max.group("max")) if m_max else None
    cur_val = m_cur.group("cur") if m_cur else None
    return pd.Series({"living_cost_min": min_val, "living_cost_max": max_val, "living_cost_currency": cur_val})


living_info = df["cost_of_living"].apply(extract_living_info)
df = pd.concat([df, living_info], axis=1)


print(df[["cost_of_living", "living_cost_min", "living_cost_max", "living_cost_currency"]].head(15).to_string(index=False))
```
